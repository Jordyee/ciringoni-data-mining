# Ciringoni · Hasil validasi Minggu ke-2

Notebook arsip ini merangkum evaluasi lokal pada 356 gambar validasi dengan 2.205 kotak acuan. Sel Python menampilkan enam hasil dan menghitung selisih berpasangan dari angka publik yang sudah dibulatkan. Notebook dapat dijalankan tanpa GPU, dataset, checkpoint, atau akses internet.

[Sumber angka dan protokol M2](https://github.com/Jordyee/ciringoni-data-mining/blob/f8d0322b2c9f3e5a598ee5828c344aba21840a1f/RINGKASAN_M2.md)

```python
from decimal import Decimal as D

# AP-small COCO, recall pada confidence 0,05 dan IoU 0,50, median CPU (ms), p95 CPU (ms).
hasil = {
    'E2': [('Native 640', '0.02189', '0.2798', '28.6', '39.2'),
           ('Native 1280', '0.04269', '0.3633', '120.4', '146.5')],
    'E3': [('Gambar utuh 640', '0.01429', '0.2467', None, None),
           ('Ubin latih 384', '0.00339', '0.1156', None, None)],
    'E4': [('Gambar penuh 1280', '0.04265', '0.3628', '120.0', '145.6'),
           ('SAHI ubin 640', '0.03899', '0.3732', '307.7', '379.9')]
}
print('Eks.  Perlakuan               AP-small     Recall     CPU median / p95 (ms)')
for eksperimen, pasangan in hasil.items():
    for perlakuan, ap, recall, median, p95 in pasangan:
        latensi = f'{median} / {p95}' if median else 'Belum tercatat'
        print(f'{eksperimen:<6}{perlakuan:<24}{ap:<13}{recall:<11}{latensi}')
```

## Perubahan terhadap kontrol masing-masing

E2 membandingkan resolusi sumber dengan lima epoch, batch 4 dan seed 42. E3 membandingkan pelatihan gambar utuh dan ubin dari data olahan 640 dengan 620 langkah optimizer per model. E4 membandingkan gambar penuh dan SAHI pada checkpoint E2 1280 yang sama. Skor E4 berasal dari GPU, sedangkan latensi berasal dari proses CPU terpisah.

```python
for eksperimen, (kontrol, kandidat) in hasil.items():
    delta_ap = D(kandidat[1]) - D(kontrol[1])
    delta_recall = D(kandidat[2]) - D(kontrol[2])
    print(f'{eksperimen}: perubahan AP-small {delta_ap:+.5f}, recall {delta_recall:+.4f}')
rasio = D(hasil['E4'][1][3]) / D(hasil['E4'][0][3])
print(f'E4: median latensi SAHI {rasio:.2f} kali gambar penuh.')
```

## Keputusan M2

E2 native 1280 dengan inferensi gambar penuh menjadi kandidat untuk tahap berikutnya. Pada E3, pelatihan ubin menurunkan AP-small. Pada E4, SAHI menaikkan recall tetapi menurunkan AP-small dan menambah latensi CPU. Seluruh kesimpulan berasal dari split validasi.
